```python
import pandas as pd

items = pd.read_csv("order_items.csv")
menu = pd.read_csv("menu.csv")
lines = items.merge(menu, on="item_id")
pizza = lines[lines["category"] == "пицца"]
revenue = (pizza["qty"] * pizza["price"]).groupby(pizza["name"]).sum()
revenue.sort_values().plot(kind="barh", title="Выручка по видам пиццы", xlabel="", ylabel="")
```
