## Урок 12.3. Проект, часть 3: проверка, график, отчёт и вывод

Пример из урока 12.3, шаг 3:

```python
import pandas as pd

days = pd.read_csv("days.csv")
days = days.assign(month=pd.to_datetime(days["date"]).dt.month, wet=days["precip_mm"] > 0)
table = days.pivot_table(index="month", columns="wet", values="orders", aggfunc="mean")
table = table.rename(columns={False: "сухо", True: "осадки"})
table.plot(title="Заказов в день: сухие и дождливые дни", xlabel="Месяц", ylabel="Заказов на город в день")
```
